```python
nums = [1,2,3]
target = 3

class Solution(object):
    def twoSum(self, nums, target):
        for i in range(len(nums)):
            for j in range(i+1, len(nums)):
                if nums[i] + nums[j] == target:
                    return[i, j]


    
```

```python
def double_numbers(nums):
    doubled = []

    for n in nums:
        doubled.append(n * 2)

    return doubled


nums = [1, 2, 3, 4, 5]

print(double_numbers(nums))

```

```python
def removed(nums):
    removed = []

    for n in nums:
        if n not in removed:
            removed.append(n)

    return removed


nums = [1, 2, 2, 3, 4, 4, 5]

print(removed(nums))
```

```python
def count(nums, target):
    sum = 0
    for n in nums:
        if n == target:
            sum = sum + 1

    return sum

nums = [2, 4, 2, 7, 2, 9]
target = 2
print(count(nums, target))
```

```python
def split(nums):
    even = []
    odd = []
    for n in nums:
        if n % 2 == 0:
            even.append(n)
        else:
            odd.append(n)

    return str(even) + " " + str(odd)

nums = [1, 2, 3, 4, 5, 6, 7]
print(split(nums))
```

```python
def numletters(word):
    counts = {}
    for n in word:
        if n in counts:
            counts[n] = counts[n] + 1
        else:
            counts[n] = 1
    return counts
        
word = "banana"
print(numletters(word))
```
